```python

#Load the dataset
df = pd.read_csv("index.csv")

#Remove duplicate rows
df = df.drop_duplicates()

#Convert date columns to numeric
date_cols = ["Year", "Month", "Day"]

for col in date_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 4. Convert indicator columns to numeric
numeric_cols = [
    "Federal Funds Target Rate",
    "Federal Funds Upper Target",
    "Federal Funds Lower Target",
    "Effective Federal Funds Rate",
    "Real GDP (Percent Change)",
    "Unemployment Rate",
    "Inflation Rate"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 5. Validate date components
valid_dates = (
    df["Year"].between(1, 9999)
    & df["Month"].between(1, 12)
    & df["Day"].between(1, 31)
)

df = df.loc[valid_dates].copy()

# 6. Create a date column and validate actual dates
df["Date"] = pd.to_datetime(
    df[["Year", "Month", "Day"]], errors="coerce"
)

df = df.dropna(subset=["Date"])

# 7. Sort chronologically
df = df.sort_values("Date")

# 8. Check missing values
print("Missing values:\n", df.isnull().sum())

# 9. Check for negative rates (flag for review)
rate_cols = [
    "Federal Funds Target Rate",
    "Federal Funds Upper Target",
    "Federal Funds Lower Target",
    "Effective Federal Funds Rate"
]

for col in rate_cols:
    print(f"Negative values in {col}:",
          (df[col] < 0).sum())

# 10. Save the cleaned dataset
df.to_csv("index_cleaned.csv", index=False)

print("Cleaning complete.")
print(df.info())
print(df.head())
```
